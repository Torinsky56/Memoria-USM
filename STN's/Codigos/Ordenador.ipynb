```python

archivo_entrada = "all_hyb_3_50.out"
archivo_salida = "all_hyb_50_ordenado.out"

with open(archivo_entrada, "r") as f:
    lineas = f.readlines()

nueva_data = []
for i in range(0, len(lineas), 3):
    grupo = lineas[i:i+3]
    if len(grupo) == 3:
        nueva_data.append(grupo[2])
        nueva_data.append(grupo[1])
        nueva_data.append(grupo[0])

with open(archivo_salida, "w") as f:
    f.writelines(nueva_data)

print(f"Archivo reordenado guardado en {archivo_salida}")
```
